```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

df = pd.read_csv('Sample - Superstore.csv', encoding='latin1')

print("Dataset Loaded Successfully!")
print("Dataset Shape:", df.shape)

df.head()
```

```python

print("Column Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())
```

```python

# Convert Order Date to datetime
df['Order Date'] = pd.to_datetime(df['Order Date'])

# Remove duplicate records
df = df.drop_duplicates()

# Handle missing values
df = df.dropna(subset=['Order Date', 'Sales'])

# Sort data by date
df = df.sort_values('Order Date')

# Check cleaned dataset
print("Data Cleaning Completed!")
print("Rows and Columns:", df.shape)
print("Missing Values:\n", df.isnull().sum())

df.head()
```

```python

monthly_sales = df.set_index('Order Date')['Sales'].resample('MS').sum()

print(monthly_sales.head())

plt.figure(figsize=(14, 6))
plt.plot(monthly_sales.index, monthly_sales.values, marker='o')
plt.title('Monthly Historical Sales Trend')
plt.xlabel('Date')
plt.ylabel('Total Sales')
plt.grid(True)
plt.show()
```

```python

# Prepare monthly sales dataset
sales_df = monthly_sales.reset_index()
sales_df.columns = ['Date', 'Sales']

# Create numerical time variable
sales_df['Month_Number'] = np.arange(len(sales_df))

# Split data into training and testing
split_index = int(len(sales_df) * 0.8)

X = sales_df[['Month_Number']]
y = sales_df['Sales']

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]
y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

# Train Linear Regression Model
model = LinearRegression()
model.fit(X_train, y_train)

# Predict test data
y_pred = model.predict(X_test)

print("Linear Regression Model Trained Successfully!")
```

```python

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("Model Evaluation Results")
print("-------------------------")
print("MAE:", round(mae, 2))
print("RMSE:", round(rmse, 2))
print("R2 Score:", round(r2, 4))
```

```python

plt.figure(figsize=(12, 6))

plt.plot(sales_df['Date'].iloc[split_index:],
         y_test, marker='o', label='Actual Sales')

plt.plot(sales_df['Date'].iloc[split_index:],
         y_pred, marker='x', linestyle='--',
         label='Predicted Sales')

plt.title('Actual vs Predicted Monthly Sales')
plt.xlabel('Date')
plt.ylabel('Sales')
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

future_months = np.arange(
    len(sales_df),
    len(sales_df) + 6
).reshape(-1, 1)

future_predictions = model.predict(future_months)

future_dates = pd.date_range(
    start=sales_df['Date'].max() + pd.offsets.MonthBegin(1),
    periods=6,
    freq='MS'
)

forecast_df = pd.DataFrame({
    'Month': future_dates,
    'Predicted Sales': np.round(future_predictions, 2)
})

print("Next 6 Months Sales Forecast")
display(forecast_df)
```

```python

forecast_df.to_csv('Superstore_Sales_Forecast.csv', index=False)

print("Forecast CSV exported successfully!")
```
